```python
df["layoff_within_12_months_flag"] = df["layoff_within_12_months_flag"].replace({
    "N": 0,
    "No": 0,
    "0": 0,
    "False": 0,
    "Y": 1,
    "Yes": 1,
    "1": 1,
    "True": 1
})
```

Cleaning "layoff_within_12_months_flag" Column By Making Value Either yes(1) Or No(0).

```python
df["visa_sponsorship_offered"] = df["visa_sponsorship_offered"].replace({
    "N": 0,
    "No": 0,
    "0": 0,
    "False": 0,
    "Y": 1,
    "Yes": 1,
    "1": 1,
    "True": 1
})
```

```python
df["job_id"].nunique()
```

```python
df.drop_duplicates(subset=["job_id"],inplace=True)
```

Dropping ALL Row Have duplicates Job_id, As Each Unique Job must have Unique Job Id.

```python
df.duplicated().sum()
```

Hence , Now There is No Duplicated Row.

```python
df["job_location_city"].isnull().sum()
```

```python
df["job_location_city"].fillna("unknown",inplace=True)
```

```python
df["primary_programming_language"].fillna("unknown",inplace=True)
```

```python
df["posting_date"]=pd.to_datetime(df["posting_date"],errors="coerce")
```

Converting posting date column's object dtype into date time.

```python
df["application_deadline"]=pd.to_datetime(df["application_deadline"],errors="coerce")
```

Converting Application_Deadlinee column's object dtype into date time.

```python
df["Deadline_Status"]=df["application_deadline"].isna().map({True:1,False:0})
```

Creating A new column Deadline Status Which Show That Dealine is given or not As 1 for deadline Given While 0 For not given.

```python
df["salary_min_usd"] = df["salary_min_usd"].astype(str)
df["salary_min_usd"] = df["salary_min_usd"].str.replace("$", "", regex=False)
df["salary_min_usd"] = df["salary_min_usd"].str.replace(",", "", regex=False)
df["salary_min_usd"] = pd.to_numeric(df["salary_min_usd"], errors="coerce")
df.loc[df["salary_min_usd"] < 0, "salary_min_usd"] = pd.NA
```

```python
df["salary_max_usd"] = df["salary_max_usd"].astype(str)
df["salary_max_usd"] = df["salary_max_usd"].str.replace("$", "", regex=False)
df["salary_max_usd"] = df["salary_max_usd"].str.replace(",", "", regex=False)
df["salary_max_usd"] = pd.to_numeric(df["salary_max_usd"], errors="coerce")
df.loc[df["salary_max_usd"] < 0, "salary_max_usd"] = pd.NA
```

Converting Object D-Type of salary_min_usd && salary_max_usd Int.

```python
df["salary_min_usd"]=df["salary_min_usd"].fillna(df.groupby("job_role")["salary_min_usd"].transform("mean"))
df["salary_max_usd"]=df["salary_max_usd"].fillna(df.groupby("job_role")["salary_max_usd"].transform("mean"))
```

The Null value of Max and Min Column is filled by the mean of min and max salary of people related to that job.

```python
df["number_of_applicants"]=df["number_of_applicants"].fillna(df.groupby("job_role")["number_of_applicants"].transform(lambda x:x.mode()[0]))
df["number_of_interview_rounds"]=df["number_of_interview_rounds"].fillna(df.groupby("job_role")["number_of_interview_rounds"].transform(lambda x:x.mode()[0]))
```

1-Filling Null value of Number_of_Applicant column with Most frequent No of people giving application for that particular job role.
2-Filling Null value of  number_of_interview_rounds column with "Most number_of_interview_rounds Belong to that particular job role.

#Feature Engineering :-

```python
df["salary_range"] = df["salary_max_usd"] - df["salary_min_usd"]
df["experience_range"] = df["years_experience_max"] - df["years_experience_min"]
```

- Creating Salary/Experience Range So it Will Help model to Predict Better .

```python
df["offer_acceptance_rate"] = df["offers_accepted"] / df["offers_extended"].replace(0, pd.NA)
```


- Offer Acceptance Rate - what % of extended offers get accepted Using replace to avoid divide-by-zero, since some rows have offers_extended = 0

```python
df["applicants_per_round"] = df["number_of_applicants"] / df["number_of_interview_rounds"].replace(0, pd.NA)
```

- Creating Column Applicants per Interview Round to show how competitive each round is.


```python
df['offer_acceptance_rate'] = df['offer_acceptance_rate'].fillna(df['offer_acceptance_rate'].mean())
df['applicants_per_round'] = df['applicants_per_round'].fillna(df['applicants_per_round'].mean())
```

```python
df["days_to_deadline"] = (df["application_deadline"] - df["posting_date"]).dt.days
df.drop(columns=["posting_date","application_deadline"], inplace=True)
```

- Days Between Posting and Deadline

```python
median_salary = df["salary_max_usd"].median()
df["is_high_salary"] = (df["salary_max_usd"] > median_salary).astype(int)
```

- HIGH Salary Flag - 1 if salary_max is above the median,else 0


```python
from sklearn.preprocessing import OrdinalEncoder
Oe=OrdinalEncoder(categories=[['Intern','Entry','Mid','Senior','Lead','Principal']])
df['seniority_level'] = Oe.fit_transform(df[['seniority_level']])
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(drop='first', sparse_output=False, dtype=np.int32)
encoded = ohe.fit_transform(df[['work_mode','company_type',"position_status"]])
encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(['work_mode','company_type',"position_status"]))
df = pd.concat([df.drop(columns=['work_mode','company_type',"position_status"]), encoded_df], axis=1)
```

```python
from sklearn.preprocessing import LabelEncoder
LB= LabelEncoder()
df['company_name'] = LB.fit_transform(df['company_name'])
df['job_role'] = LB.fit_transform(df['job_role'])
df['sector'] = LB.fit_transform(df['sector'])
df['hq_country'] = LB.fit_transform(df['hq_country'])
df['job_location_city'] = LB.fit_transform(df['job_location_city'])
df['job_location_country'] = LB.fit_transform(df['job_location_country'])
df['primary_programming_language'] = LB.fit_transform(df['primary_programming_language'])
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df[['salary_min_usd','salary_max_usd','years_experience_min','years_experience_max',
    'company_size_employees','number_of_applicants','number_of_interview_rounds',
    'offers_extended','offers_accepted','experience_range','offer_acceptance_rate',
    'applicants_per_round','days_to_deadline']] = scaler.fit_transform(
    df[['salary_min_usd','salary_max_usd','years_experience_min','years_experience_max',
        'company_size_employees','number_of_applicants','number_of_interview_rounds',
        'offers_extended','offers_accepted','experience_range','offer_acceptance_rate',
        'applicants_per_round','days_to_deadline']])
```
