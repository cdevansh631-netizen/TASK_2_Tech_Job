The Null value of Max and Min Column is filled by the mean of min and max salary of people related to that job.

1-Filling Null value of Number_of_Applicant column with Most frequent No of people giving application for that particular job role.

#Feature Engineering :-

```python
df["Avg_Salary"]=(df["salary_min_usd"]+df["salary_max_usd"])/2
df["experience_range"] = df["years_experience_max"] - df["years_experience_min"]
```

- Creating Salary/Experience Range So it Will Help model to Predict Better .

```python
df["offer_acceptance_rate"] = df["offers_accepted"] / df["offers_extended"].replace(0, pd.NA)
```


- Offer Acceptance Rate - what % of extended offers get accepted Using replace to avoid divide-by-zero, since some rows have offers_extended = 0

```python
df["applicants_per_round"] = df["number_of_applicants"] / df["number_of_interview_rounds"].replace(0, pd.NA)
```

- Creating Column Applicants per Interview Round to show how competitive each round is.


```python
df['offer_acceptance_rate'] = df['offer_acceptance_rate'].fillna(df['offer_acceptance_rate'].mean())
df['applicants_per_round'] = df['applicants_per_round'].fillna(df['applicants_per_round'].mean())
```

```python
df["days_to_deadline"] = (df["application_deadline"] - df["posting_date"]).dt.days
df.drop(columns=["posting_date","application_deadline"], inplace=True)
```

- Days Between Posting and Deadline

```python
median_salary = df["salary_max_usd"].median()
df["is_high_salary"] = (df["salary_max_usd"] > median_salary).astype(int)
```

```python
df.dropna(subset=["offers_extended","offers_accepted"],inplace=True)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
X_train.isnull().sum().sort_values(ascending=False).head(5)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,y_train)
```

- Linear Regression use a ols-(ordinary test sequence) method to minimize  mean squared error value and finds best value of weight/parameter and bias  with help of linear formula Hence It didn't use gradient descent which use iterative approach to find best parameter Values.

- Therefore Reduce Cost Function as Minimum It can do in linear regression.

```python
y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import root_mean_squared_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

print("Mean Absolute Error : ",mean_absolute_error(y_test,y_pred))
print("Root Mean Squared Error : ",root_mean_squared_error(y_test,y_pred))
print("Mean Squared Error : ",mean_squared_error(y_test,y_pred))
print("R2 Score : ",r2_score(y_test,y_pred))
```

- Mean Absolute Error Shows us the average/absolute error occur during prediction.Hence On An Average actual and predicted salary differ by 21546.

- Root Mean Squared Error is square root of average squared error.As it Give Extra importance to those whose error value is large as it is squared root error.

- Root mean sqared Error and Mean Absolute error has a diff that MAE trat all error normally While RMSE give extra importance to RMSE.


- MSE is mean of sum of squared error occurs.

- R2 Score How much variation on Actual Value is explained by our model.Here 0.78 r2 score mean Model explain 78% of variation in salary.

```python
M=max(y_test.max(),y_pred.max())
L=min(y_test.min(),y_pred.min())
plt.scatter(y_test,y_pred,alpha=.64)
plt.plot([M,L],[M,L],color="red")
```

- Model is Good For prediction For the typical Salary But Not Good for extreme outliers.
